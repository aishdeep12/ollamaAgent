```python
import pandas as pd 
from sklearn.preprocessing import LabelEncoder
from sklearn.tree  import DecisionTreeClassifier,export_text
from sklearn.ensemble import RandomForestClassifier
```

```python
data = {
    "Age": ["Young","Young","Middle","Old","Old","Old","Middle","Young","Young","Old","Young","Middle","Middle","Old"],
    "Income": ["High","High","High","Medium","Low","Low","Low","Medium","Low","Medium","Medium","Medium","High","Medium"],
    "Student": ["No","No","No","No","Yes","Yes","Yes","No","Yes","Yes","Yes","No","Yes","No"],
    "Credit": ["Fair","Good","Fair","Fair","Fair","Good","Good","Fair","Fair","Fair","Good","Good","Fair","Good"],
    "Buy": ["No","No","Yes","Yes","Yes","No","Yes","No","Yes","Yes","Yes","Yes","Yes","No"]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
encoders = {}
```

```python
for col in df.columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    encoders[col] = le
```

```python
X = df[["Age","Income","Student","Credit"]]
Y = df["Buy"]
```

```python
dt = DecisionTreeClassifier(random_state = 42 , max_depth = 3)
```

```python
dt.fit(X,Y)
```

```python
rf = RandomForestClassifier(n_estimators = 5, max_depth = 3, max_features = 2,random_state = 42)
```

```python
rf.fit(X,Y)
```

```python
new_people = pd.DataFrame([
    {"Age": "Young", "Income": "High", "Student": "No", "Credit": "Fair"},
    {"Age": "Young", "Income": "Low", "Student": "Yes", "Credit": "Good"},
    {"Age": "Old", "Income": "Medium", "Student": "No", "Credit": "Good"}
])
new_people
```

```python
for col in new_people.columns:
    new_people[col] = encoders.get(col).transform(new_people[col])
```

```python
new_people
```

```python
dt_prediction = encoders["Buy"].inverse_transform(dt.predict(new_people))
```
